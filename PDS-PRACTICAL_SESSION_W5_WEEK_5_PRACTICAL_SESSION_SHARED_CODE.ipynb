```python
# ============================================================
# DATA STRUCTURES - STACKS & QUEUES
# ============================================================
# ============================================================
# ACTIVITY 1: BASIC STACK
# LIFO = Last In, First Out
# ============================================================
stack = []
# Add items
stack.append("A")
stack.append("B")
stack.append("C")
print("Stack:", stack)
```

```python
# See the top item
print("Top item:", stack[-1])
```

```python
# Remove the top item
removed = stack.pop()

print("Removed:", removed)
print("Stack after pop:", stack)
```

```python
# ============================================================
# ACTIVITY 3: UNDO MECHANISM
# Stack is ideal for Undo because of LIFO
# ============================================================
text = ""
undo_stack = []
```

```python
text = ""
undo_stack = []

def type_text(new_text):
    global text

    # Save current text before making a change
    undo_stack.append(text)

    # Add new text
    text = text + new_text

    print("Text:", text)


def undo():
    global text

    if undo_stack:
        text = undo_stack.pop()
        print("Undo:", text)
    else:
        print("Nothing to undo")


# Try the editor
type_text("Hello ")
type_text("World")
type_text("!")

print("\nPerforming Undo operations:")
undo()
undo()
undo()
undo()
```

```python
# ============================================================
# ACTIVITY 5: BROWSER BACK BUTTON
# Stack example
# ============================================================

history = []
history.append("Google")
history.append("YouTube")
history.append("Instagram")
history.append("LinkedIn")

print("\nBrowser History:", history)
print("Current page:", history[-1])

# Back button
history.pop()

print("After Back:", history)
print("Current page:", history[-1])
```
