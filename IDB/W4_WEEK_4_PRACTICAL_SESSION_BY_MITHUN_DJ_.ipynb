```python
marks = [78,98,56,29,66,92,88,65]
```

```python
# Remove a value
marks.remove(66)
print("After remove: ", marks)
```

```python
# remove last items
marks.pop()
print("After pop:", marks)
```

```python
# sort
marks.sort()
print("Sorted marks:", marks)
```

```python
# reverse sort
marks.sort(reverse=False)

print("Ascending sorted marks:", marks)
```

```python
# reverse sort
marks.sort(reverse=True)

print("Descending sorted marks:", marks)
```

```python
# length
print("Length of marks:", len(marks))
```

```python
# total and average
print("Total:", sum(marks))
```
